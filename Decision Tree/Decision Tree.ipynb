# Load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Social_Network_Ads.csv")
df
```

# Features & Target

```python
X = df.iloc[:,[2,3]].values
y = df.iloc[:,-1]
```

```python
plt.scatter(X[y==0,0],X[y==0,1],label="Not purchased",color="red")
plt.scatter(X[y==1,0],X[y==1,1],label="purchased",color="green")
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.legend()
plt.title("Relation between Age & Estimated salary")
plt.show()
```

```python
x = df.iloc[:,[2,3]]
x.describe()
```

# Standerization (Not required)

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
```

# Train-Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.75,random_state=0)
```

# Model Creation

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
```

# Train acc, Test acc

```python
from sklearn.metrics import confusion_matrix,accuracy_score
y_pred = model.predict(X_test)

test_acc = accuracy_score(y_test,y_pred)
test_acc
```

```python
y_pred1 = model.predict(X_train)
train_acc = accuracy_score(y_pred1,y_train)
train_acc
```

```python
#Visualising the testng set results
from matplotlib.colors import ListedColormap
X_set, y_set = X_test, y_test

X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min()-1, stop = X_set[:,0].max()+1, step = 0.01),
                     np.arange(start = X_set[:, 1].min()-1, stop = X_set[:,1].max()+1, step = 0.01))

plt.contourf(X1, X2, model.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
                                        alpha = 0.75, cmap = ListedColormap(('red', 'green')))


plt.scatter(X_set[y_set==0,0],X_set[y_set==0,1],label="Not purchased",color="red")
plt.scatter(X_set[y_set==1,0],X_set[y_set==1,1],label="purchased",color="green")
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.legend()
plt.title("Relation between Age & Estimated salary")
plt.show()
```

# Discuss Overfitting issues

# Pruning

```python
#DecisionTreeClassifier(criterion="gini",max_depth=6,ccp_alpha=0.0)
#Grid search : Hyperparameter Tuning
```

```python
model1 = DecisionTreeClassifier()
path = model1.cost_complexity_pruning_path(X_train,y_train)
```

```python
ccp_alpha,imp = path.ccp_alphas,path.impurities
```

```python
ccp_alpha
```

```python
clfs = []
for i in ccp_alpha:
    model1= DecisionTreeClassifier(ccp_alpha=i)
    clfs.append(model1)
```

```python
train_score = []
test_score = []
for mod in clfs:
    mod.fit(X_train,y_train)
    train_score.append(mod.score(X_train,y_train))
    test_score.append(mod.score(X_test,y_test))
```

```python
train_score,test_score
```

```python
plt.scatter(ccp_alpha,train_score,color="green",label="Train")
plt.plot(ccp_alpha,train_score,color="green",label="Train")
plt.scatter(ccp_alpha,test_score,color='red',label="test")
plt.plot(ccp_alpha,test_score,color='red',label="test")
plt.grid()
plt.legend()
plt.xlabel("CCP alpha")
plt.ylabel("Accuracy")
plt.xlim(0,0.0040)
plt.show()
```

# Final Model

```python
from sklearn.tree import DecisionTreeClassifier
final_model = DecisionTreeClassifier(ccp_alpha=0.00384921)
final_model.fit(X_train,y_train)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
y_pred = final_model.predict(X_test)
test_acc = accuracy_score(y_test,y_pred)
test_acc
```

```python
y_pred1 = final_model.predict(X_train)
train_acc = accuracy_score(y_pred1,y_train)
train_acc
```

```python
#Visualising the testng set results
from matplotlib.colors import ListedColormap
X_set, y_set = X_test, y_test

X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min()-1, stop = X_set[:,0].max()+1, step = 0.01),
                     np.arange(start = X_set[:, 1].min()-1, stop = X_set[:,1].max()+1, step = 0.01))

plt.contourf(X1, X2, final_model.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
                                        alpha = 0.75, cmap = ListedColormap(('red', 'green')))


plt.scatter(X_set[y_set==0,0],X_set[y_set==0,1],label="Not purchased",color="red")
plt.scatter(X_set[y_set==1,0],X_set[y_set==1,1],label="purchased",color="green")
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.legend()
plt.title("Relation between Age & Estimated salary")
plt.show()
```
